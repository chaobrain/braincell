# SC to DBNN

Interactive entry point for the canonical SC-to-DBNN workflow. The Python module owns teacher generation, data splitting, training, calibration, and artifact persistence.


```python
from argparse import Namespace
from pathlib import Path
import sys

def find_repo_root(start):
    for candidate in (start, *start.parents):
        if (candidate / 'braincell').is_dir() and (candidate / 'examples').is_dir():
            return candidate
    raise FileNotFoundError('Could not locate the BrainCell repository root.')

REPO_ROOT = find_repo_root(Path.cwd().resolve())
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from examples.reduction.sc_to_dbnn import PROFILES, WORKFLOW_VERSION, run
```

```python
PROFILE = 'full'
OUTPUT_ROOT = REPO_ROOT / 'artifacts' / 'reduction' / 'sc_to_dbnn' / f'v{WORKFLOW_VERSION}'
args = Namespace(profile=PROFILE, output_dir=OUTPUT_ROOT)
PROFILES[PROFILE]
```

```python
summary = run(args)
summary
```
